```python

df_raw = pd.read_csv(r"C:\Users\santo\OneDrive\Desktop\Real estate intelligence\data\raw\real_estate_market_in_India.csv")

print("Original dataset loaded successfully!")
print("Number of rows:", df_raw.shape[0])
print("Number of columns:", df_raw.shape[1])

df_raw.head()
```

```python

df = df_raw.copy()

print("Working copy created.")
print("Original rows:", len(df_raw))
print("Working-copy rows:", len(df))
```

```python

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nTotal missing cells:", df.isnull().sum().sum())
```

```python

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
print("Rows before duplicate removal:", len(df))
```

```python
df[df["SQUARE_FT"] > 100000]
```

```python

print("Area statistics:")
print(df["SQUARE_FT"].describe())

print("\nSmallest 10 area values:")
print(df["SQUARE_FT"].nsmallest(10))

print("\nLargest 10 area values:")
print(df["SQUARE_FT"].nlargest(10))
```

```python

price_column = "TARGET(PRICE_IN_LACS)"

print("Price statistics:")
print(df[price_column].describe())

print("\nRecords with the 10 largest reported areas:")
df.nlargest(10, "SQUARE_FT")[
    ["SQUARE_FT", price_column, "BHK_NO.", "ADDRESS"]
]
```
